```python
#using function

def func(param: list[str])->  list[str]:
 param.reverse()
 return param

print(func(["1","2"]))
```

```python
def func1(par: str):
    par=par[::-1]
    return par

print(func1("github"))
```

```python
def fun(first_name: str ,last_name: str) -> str:
    return first_name.upper() + " " + last_name.upper()
    

print(fun("sanket","kolte"))
```

```python
def f(data:tuple[int,str,int]):
    return data

print(f(["sanket",10,40]))
```

```python
#using union "|"pipe operator

def function(item: int | str):
    print(item,type(item))

function(100)
function("sanket")
function(True)
```

```python
#iterable type use when we dont know the what type of data type is sent to function 

from collections.abc import Iterable 

def func2(item: Iterable[str]) -> tuple[str]:
    return item[::-1]

print(func2(("1","2")))
```

```python
### typehint use with class 

class Person:
    def __init__(self,name: str):
        self.name=name

def check(var: Person) -> str:
    return var.name

p=Person("sanket")

print(check(p))
```

```python
# using Literal in typehint

from typing import Literal

def check(name: str,role : Literal["admin","emp","viwer"]) -> str:
    return f"{name} has a {role} permission"

print(check("sanket","admin"))
print(check("sham","viwer"))
print(check("ram","xyz"))

```

```python
#use Enum in typehint

from enum import Enum

class Role(Enum):
    ADMIN = "adimn"  #ADMIN is use name and "admin to value"
    VIEWER = "viewer"
    EDITOR = "editor"

def acess(user: str , role : Role)-> str:
    return f"{user} has role name = {role.name} and role value ={role.value}"

print(acess("ram",Role.EDITOR))
```
