```python
balance=5000
try:
    amount=int(input("enter the amount"))
    if amount < 0:
        raise ZeroDivisionError("amount must be gerater than zero")
    elif amount > balance:
        raise ValueError("insufficient balance")
except ZeroDivisionError as e:
    print(e)
except ValueError as e:
    print(e)
except Exception as e:
    print(e)
else:
    balance -= amount
    print("amount withdraw sucessfully")
finally:
    print("thank you for visti")
```
